```python
# Remove duplicates

df.drop_duplicates(inplace=True)
```

```python
# Replacing invalide
df.replace(['N/A','NULL',' '],np.nan,inplace=True)
```

```python
df
```

```python
# Removing extra scapes

df= df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
#Standertizing the text format
# Converting text into proper case

df['Customer_Name']=df['Customer_Name'].str.title()
df['city']=df['city'].str.title()
df['state']=df['state'].str.title()
```

```python
# Validating email address
# Keeping only valid email records

df= df[df['Email'].str.contains('@',na=False)]
```

```python
# Converting data columns

df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
#Converting numeric columns

cols = ['Qty','Discount','Unit_Price']
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')
```

```python
# Removing invalid quantities
# remove records having zero or negative values

df=df[df['Qty']>0]
```

```python
# Filling the missing values

df['Discount'].fillna(0,inplace=True)
df['Phone'].fillna('Unknown',inplace=True)
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
df
```

##FEATURE ENGINERING

```python
#Total Sales (adding column in table)

df['Sales']=df['Qty']*df['Unit_Price']
```

```python
df
```

```python
#Net Amount

df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
# Profit (ex:-20%)
df['Profit']=df['Net_Amount']*0.20
```

```python
# Order Month
df['Month']=df['Order_Date'].dt.month_name()
```

```python
#Order Year
df['Year']=df['Order_Date'].dt.year
```

```python
# Weekday

df['Weekday']=df['Order_Date'].dt.day_name()
```
